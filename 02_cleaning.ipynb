```python
import ast
df["Цена"] = (
    df["Цена"]
    .astype(str)
    .str.replace(" ", "")                 
    .str.replace(r"[^\d]", "", regex=True)  
)

df["Цена"] = pd.to_numeric(df["Цена"], errors="coerce").astype("Int64")

df["Описание"] = df["Описание"].astype(str).str.replace(r"^Описание\\n?", "", regex=True).str.strip()

df["Категории"] = df["Категории"].apply(lambda x: ast.literal_eval(x) if pd.notnull(x) and x.startswith("[") else [])

df["Аптеки"] = df["Аптеки"].apply(lambda x: ast.literal_eval(x) if pd.notnull(x) and x.startswith("[") else [])

print(df.dtypes)
```

```python
df["Категории"] = df["Категории"].apply(lambda x: x[2:] if isinstance(x, list) and x[:2] == ["Главная", "Каталог"] else x)
```

```python
import re

def clean_apteka(entry):
    if not isinstance(entry, str):
        return None
    entry = re.sub(r"\([^)]*\)", "", entry)
    entry = re.sub(r"[\(\)]", "", entry)

    entry = re.sub(r"\bг\.?\s*Алматы\b", "", entry, flags=re.IGNORECASE)
    entry = re.sub(r"\bАлматы\b", "", entry, flags=re.IGNORECASE)

    entry = re.sub(r"\b[а-яА-Яё\-]*\s?район\b", "", entry, flags=re.IGNORECASE)
    entry = re.sub(r"\bр[\.\-]?[а-яА-Яё0-9]*\b", "", entry, flags=re.IGNORECASE)

    entry = re.sub(r"\b[А-ЯЁ][а-яё]+ский\b", "", entry)

    entry = re.sub(r",\s*(ул\.|улица|мкр|проспект|бульвар|переулок|жк|корпус|д\.)", r" \1", entry, flags=re.IGNORECASE)

    entry = re.sub(r"\s{2,}", " ", entry)
    entry = re.sub(r",\s*,", ",", entry)
    entry = re.sub(r"\s*,\s*", ", ", entry)
    entry = entry.strip(" ,.")

    return entry
```

```python
def clean_description(text):
    if not isinstance(text, str):
        return "Описание отсутствует"
    
    text = text.strip()
    
    text = re.sub(r"^Описание\\n?", "", text, flags=re.IGNORECASE)
    text = re.sub(r"^Описание\n?", "", text, flags=re.IGNORECASE)

    if text.strip().lower() == "нет":
        return "Описание отсутствует"

    text = re.sub(r"\s{2,}", " ", text)
    text = text.replace("\n", " ").strip(" ,.")

    return text

df["Описание"] = df["Описание"].apply(clean_description)
```
